```python
import numpy as np
import matplotlib.pyplot as plt

n_paths = 500
n_steps = 100
T = 1.0
dt = T / n_steps
Sqrtdt=np.sqrt(dt)
rng = np.random.default_rng() #initialize rand numb generator
X = np.zeros((n_paths, n_steps+1)) #Allocate a zero array of trajectories

for i in range(n_paths):
    for j in range(1,n_steps+1):
    # Assign a value based on the row (i) and column (j) index
        X[i, j] = X[i, j-1]+rng.normal(loc=0, scale=1)*Sqrtdt

# Time axis (n_steps+1) steps from 0 to T
t = np.linspace(0, T, n_steps + 1)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(t, X.T, alpha=0.5, color='blue') 
plt.title('500 Brownian Motion Trajectories')
plt.xlabel('Time t')
plt.ylabel('X(t)')
plt.grid(True)
plt.show()
#X.T is transposed X
```
